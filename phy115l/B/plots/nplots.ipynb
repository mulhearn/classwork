```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
A = 1.0
B = 0.1

def ws_pot(x):
    return -A/(1+np.exp((x-1)/B))

x = np.arange(0,2,0.01)
ya = ws_pot(x)
plt.plot(x,ya,"b-",label="signal")
plt.yticks([-1,0],["$-V_0$","0"])
plt.xticks([0,1,2],["0","R","2R"])
plt.xlabel("$r$")
plt.ylabel("$V(r)$")

plt.savefig("woodssaxon.pdf")
```

```python
x = np.arange(0,2,0.01)
ya = np.cos(2*np.pi*x)
yb = x*0
plt.plot(x,ya,"b-",label="signal")
plt.plot(x,yb,"r--")
plt.xlim(0,1)
xS=np.array([0,0.5,1.0])
yS=np.array([1.0,-1.0,1.0])
plt.plot(xS,yS,"ko",label="samples")
plt.xlabel("$x$");
plt.ylabel("$\psi(x)$");
plt.legend()

plt.text(0.7,0.2,"$a/N$")

xL=np.array([0.5,1.0])
yL=np.array([0.15,0.15])
plt.plot(xL,yL,"k-|")
plt.savefig("notrouble.pdf")
```
