```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
plt.rcParams.update({'font.size': 12})
```

```python
x = np.linspace(0,1,200)
y = x**2
plt.plot(x,y)
plt.xlabel("T")
plt.ylabel("$P_{fi}$")
plt.xticks([0,1],["",""])
plt.yticks([0,1],["",""])
plt.text(0.5,0.5,"$P_{fi} \propto T^2$")
plt.savefig("pfi_quad.pdf")
```

```python
x = np.linspace(0,1,200)
y = x**2
plt.plot(x,np.sin(2*np.pi*x)**2)
plt.xlabel("T")
plt.ylabel("$P_{fi}$")
plt.xticks([0,0.5,1.0],["0","$2 \pi / \omega_{fi}$","$4 \pi / \omega_{fi}$"])
plt.yticks([0,1],["0",""])
plt.axhline(y=1,color="k",linestyle="--")
plt.text(0.35,0.93,"$4 |V_{fi}|^2 / (E_f - E_i)^2$")
plt.savefig("pfi_osc.pdf")
```

```python


x = np.linspace(-1,1,200)
plt.plot(x,np.sinc(2*x)**2)
plt.xlabel("$\omega_{fi}$")
plt.ylabel("$F_T(\omega_{fi})$")
plt.xticks([-1, -0.5, 0,0.5,1.0],["$-4 \pi / T$","$-2 \pi / T$", "0","$2 \pi / T$","$4 \pi / T$"])
plt.yticks([0,1],["0",""])
plt.axhline(y=1,color="k",linestyle="--")
plt.text(-0.5,0.93,"$T^2$")
plt.savefig("fgr_delta.pdf")
```
